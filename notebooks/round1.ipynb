```python
from borrel.policies.dummy import Dummy
from borrel.policies.ivo import Ivo
from borrel.policies.jakko import Jakko
from borrel.policies.mark import Mark
from borrel.policies.allia import Allia
from borrel.policies.dominique import Dominique
from borrel.policies.dirk import Dirk
from borrel.policies.mees import Mees
from borrel.policies.daniel import Daniel

from borrel.games.tron import Tron
from borrel.games.sorry import Sorry
from borrel.games.dots_and_lines import DotsAndLines
from borrel.games.rps_gun import RPS_Gun

from itertools import combinations

def get_all_player_classes():
    """
    Here we define all the players that are available to be used in the tournament.
    """
    return [
        Ivo,
        Mark,
        Dominique,
        Dirk,
        Mees,
        Daniel,
    ]

def init_all_players():
    return [
        player() for player in get_all_player_classes()
    ]

def get_zero_scores(players):
    return {
        player.name: 0
        for player in players
    }

global_scores = get_zero_scores(init_all_players())

def print_ordered_scores(scores: dict[str, int]):
    """
    Ranks the scores dict by score
    """
    ordered_scores = sorted(scores.items(), key=lambda item: item[1], reverse=True)
    print("Scores: " + ", ".join([f"{name}: {score}" for name, score in ordered_scores]))

def print_global_scores():
    ordered_scores = sorted(global_scores.items(), key=lambda item: item[1], reverse=True)
    print("Global scores:")
    for name, score in ordered_scores:
        print(f"{name}: {score}")

def update_global_scores(scores: dict[str, int]):
    """
    Ranks the scores dict by score
    """
    sorted_scores = sorted(scores.items(), key=lambda item: item[1], reverse=True)
    last_score = None
    current_points = len(init_all_players()) - 1
    for name, score in sorted_scores:
        if last_score is not None and last_score != score:
            current_points -= 1
        global_scores[name] += current_points

        last_score = score

    print_global_scores()
```

## Tron

```python
scores = get_zero_scores(init_all_players())

player_classes = get_all_player_classes()

player_classes = [player_class for player_class in player_classes if player_class not in [Mark]]

for player1_class, player2_class in combinations(player_classes, 2):
    for i in range(11):
        player1 = player1_class()
        player2 = player2_class()

        tron = Tron([player1, player2], render=False)
        winner = tron.play_game()
        if winner == "tie":
            continue

        scores[winner] += 1

print_ordered_scores(scores)

update_global_scores(scores)
```

## Sorry

```python
scores = get_zero_scores(init_all_players())

player_classes = get_all_player_classes()

player_classes = [player_class for player_class in player_classes if player_class not in [Dirk, Daniel, Mark]]

for i in range(11):
    sorry = Sorry(players=[player() for player in player_classes], render=False)
    winners = sorry.play_game()

    print(f"Winner(s): {winners}")

    for winner in winners:
         scores[winner] += 1

print_ordered_scores(scores)

update_global_scores(scores)
```

## Dots and Lines

```python
scores = get_zero_scores(init_all_players())

player_classes = get_all_player_classes()

for player1_class, player2_class in combinations(player_classes, 2):
    for i in range(11):
        player1 = player1_class()
        player2 = player2_class()

        dots_and_lines = DotsAndLines(player1, player2, size=5, render=False)
        winner = dots_and_lines.play_game()

        print(dots_and_lines.scores, f", Winner: {winner}")

        if winner == "tie":
            continue

        scores[winner] += 1

print_ordered_scores(scores)

update_global_scores(scores)
```

```python
scores = get_zero_scores(init_all_players())

player_classes = get_all_player_classes()

player_classes = [player_class for player_class in player_classes if player_class not in [Mark]]

for player1_class, player2_class in combinations(player_classes, 2):
    for i in range(5):
        player1 = player1_class()
        player2 = player2_class()

        rps_gun = RPS_Gun(player1, player2, render=False)
        winner = rps_gun.play_game()

        print(rps_gun.scores, f", Winner: {winner}")

        if winner == "tie":
            continue

        scores[winner] += 1

print_ordered_scores(scores)

update_global_scores(scores)
```
