Here you can test your stuff.

For each of the games I have preset:
- A single game with rendering/printing enabled so you can see what's going on
- A competition length game against the dummy player to benchmark your performance
    - Please keep the runtime of this cell (when competing against dummy) under like ~45 seconds.

After the first scoring round you can (and probably should!) test your policies against the other players!

Players

```python
from borrel.policies.dummy import Dummy
from borrel.policies.ivo import Ivo
from borrel.policies.jakko import Jakko
from borrel.policies.mark import Mark
from borrel.policies.allia import Allia
from borrel.policies.dominique import Dominique
from borrel.policies.dirk import Dirk
from borrel.policies.mees import Mees
from borrel.policies.daniel import Daniel

from borrel.games.tron import Tron
from borrel.games.minority import Minority
from borrel.games.sorry import Sorry
from borrel.games.dots_and_lines import DotsAndLines
from borrel.games.rps_gun import RPS_Gun

you_class = Dummy # Change this to your class :)
```

## Minority

Single game


```python
# Init players
you = you_class()
opponent1 = Dummy(name="dummy1")
opponent2 = Dummy(name="dummy2")
opponent3 = Dummy(name="dummy3")

players = [you, opponent1, opponent2, opponent3]

minority = Minority(players, render=True)

winner = minority.play_game(rounds = 5)

print(f"Winner: {winner}")
print(f"Scores: {minority.get_scores()}")
print("History:")
minority.history
```

Benchmark

```python
you = you_class()
opponent1 = Dummy(name="dummy1")
opponent2 = Dummy(name="dummy2")
opponent3 = Dummy(name="dummy3")

scores = {you.name: 0, opponent1.name: 0, opponent2.name: 0, opponent3.name: 0}

for i in range(21):
    you = you_class()
    opponent1 = Dummy(name="dummy1")
    opponent2 = Dummy(name="dummy2")
    opponent3 = Dummy(name="dummy3")
    players = [you, opponent1, opponent2, opponent3]

    minority = Minority(players, render=False)

    winners = minority.play_game()

    for winner in winners:
        scores[winner] += 1

    print(minority.get_scores(), f"winner: {winners}")

print(f"Scores: {scores}")
```

## Tron

Single game

```python
# Note - you start as the left player ("-"). Switch the order of players to play as the right player ("|").

you = you_class()
opponent = Dummy()

tron = Tron([you, opponent], render=True)
winner = tron.play_game()

print(f"Winner: {winner}")
print("Final grid:")
tron.renderer.render()
```

Benchmark

```python
you = you_class()
opponent = Dummy()
score = {you.name: 0, opponent.name: 0}

for i in range(11):
    you = you_class()
    opponent = Dummy()

    tron = Tron([you, opponent], render=False)
    winner = tron.play_game()
    if winner == "tie":
        continue
    score[winner] += 1

print(f"Scores: {score}")
```

## Rock Paper Scissors Gun Duck

Single game

```python
you = you_class()
opponent = Dummy()

rps_gun = RPS_Gun(you, opponent, render=True)

winner = rps_gun.play_game(rounds=5)
print(f"Final scores: {rps_gun.scores}")
rps_gun.history
```

Benchmark

```python
you = you_class()
opponent = Dummy()

scores = {you.name: 0, opponent.name: 0}

for i in range(7):
    you = you_class()
    opponent = Dummy()

    rps_gun = RPS_Gun(you, opponent, render=False)
    winner = rps_gun.play_game()
    if winner == "tie":
        continue
    scores[winner] += 1
    print(rps_gun.scores, f"winner: {winner}")

print(f"Scores: {scores}")
```

## Dots and Lines

Single game

```python
# Init players
player1 = you_class()
player2 = Dummy(name="loser")

dots_and_lines = DotsAndLines(player1, player2, size=3, render=True)
winner = dots_and_lines.play_game()

print("Winner:", winner)
print("Scores:", dots_and_lines.scores)
print("Turns:", len(dots_and_lines.history))
```

Competition-length benchmark vs dummy

```python
wins = {"Dummy_1": 0, "Dummy_2": 0, "tie": 0}

for _ in range(100):
    player1 = Dummy("Dummy_1")
    player2 = Dummy("Dummy_2")

    dots_and_lines = DotsAndLines(player1, player2, size=5, render=False)
    winner = dots_and_lines.play_game()
    wins[winner] += 1

wins
```
