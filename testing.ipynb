Here you can test your stuff.

For each of the games I have preset:
- A single game with rendering/printing enabled so you can see what's going on
- A competition length game against the dummy player to benchmark your performance

After the first scoring round you can (and probably should!) test your policies against the other players! 

Players

```python
from borrel.games.sorry import Sorry
from borrel.games.dots_and_lines import DotsAndLines
from borrel.games.rps_gun import RPS_Gun
from borrel.policies.dummy import Dummy

you_class = Dummy # Change this to your own policy class
```

## Sorry

Single game

```python
# Init players
players = [
    you_class(),
    Dummy("Dummy_2"),
    Dummy("Dummy_3"),
    Dummy("Dummy_4"),
]

sorry = Sorry(players, render=True)

sorry.play_game(n_games=1)
```

## Dots and Lines

Single game

```python
# Init players
player1 = Dummy("Dummy_1")
player2 = Dummy("Dummy_2")

dots_and_lines = DotsAndLines(player1, player2, size=3, render=True)
winner = dots_and_lines.play_game()

print("Winner:", winner)
print("Scores:", dots_and_lines.scores)
print("Turns:", len(dots_and_lines.history))
```

Competition-length benchmark vs dummy

```python
wins = {"Dummy_1": 0, "Dummy_2": 0, "tie": 0}

for _ in range(100):
    player1 = Dummy("Dummy_1")
    player2 = Dummy("Dummy_2")

    dots_and_lines = DotsAndLines(player1, player2, size=3, render=False)
    winner = dots_and_lines.play_game()
    wins[winner] += 1

wins
```

## ROCK PAPER SCISSORS GUN DUCK

Single game

```python
you = you_class()
opponent = Dummy("Opponent")

rps_gun = RPS_Gun(you, opponent, render=True)

winner = rps_gun.play_game(rounds=5)

print("Winner:", winner)
```

Competition-length benchmark vs dummy

```python
you = you_class()
opponent = Dummy("Opponent")

scores = {you.name: 0, opponent.name: 0}

for i in range(7):
    you = you_class()
    opponent = Dummy("Opponent")

    rps_gun = RPS_Gun(you, opponent, render=True)
    winner = rps_gun.play_game()
    if winner == "tie":
        continue
    scores[winner] += 1
    print(f"winner: {winner}")

print(f"Scores: {scores}")
```
